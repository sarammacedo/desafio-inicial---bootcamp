QUESTÃO 1.1:

a) A média é a soma de todos os valores, divididos pelo número de elementos, sendo apropriado usar quando os dados são uniformemente distribuídos. Mediana é o valor central de um conjunto ordenado, sendo apropriado usar quando existem outliers que  podem distorcer a média. Moda é o valor que mais se repete, sendo mais apropriado usar quando se quer saber o valor mais frequente.

b) O desvio padrão é uma medida que indica o grau de dispersão dos dados, o quanto os valores de um conjunto variam. É importante na análise de dados pois ajuda a entender a consistência dos mesmos, essencial para comparações estatísticas e para definir intervalos de confiança e distribuições normais. Sua relação com a variância: desvio padrão = raiz quadrada da variância.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# DataFrame
dados = {
 'Estudante': ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena', 'Felipe', 'Gina', 'Hugo', 'Igor', 'Julia'],
 'Matematica': [85, 92, 78, 65, 88, 72, 95, 81, 69, 90],
 'Portugues': [78, 85, 92, 70, 82, 75, 88, 79, 83, 91],
 'Ciencia': [82, 88, 75, 68, 90, 71, 94, 76, 80, 87]
}
df = pd.DataFrame(dados)

# a) Estatísticas
estatisticas = df[['Matematica', 'Portugues', 'Ciencia']].agg(['mean', 'median', 'std'])
print("=== Estatísticas (Média, Mediana e Desvio Padrão) ===")
print(estatisticas)

correlacao = df['Matematica'].corr(df['Portugues'])
print("\n=== Correlação Matemática x Português ===")
print(correlacao)

# b) Visualizações
sns.set(style="whitegrid")

# Boxplot comparando as três matérias
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['Matematica', 'Portugues', 'Ciencia']])
plt.title("Boxplot das Notas por Matéria")
plt.show()

# Histograma das notas de Matemática
plt.figure(figsize=(8,5))
sns.histplot(df['Matematica'], bins=6, kde=True, color='blue')
plt.title("Histograma das Notas de Matemática")
plt.xlabel("Notas de Matemática")
plt.ylabel("Frequência")
plt.show()

# Gráfico de dispersão entre Matemática e Ciência
plt.figure(figsize=(8,5))
sns.scatterplot(x='Matematica', y='Ciencia', data=df, s=80, color='green')
plt.title("Dispersão: Matemática x Ciência")
plt.xlabel("Notas de Matemática")
plt.ylabel("Notas de Ciência")
plt.show()
```

a) Um HISTOGRAMA é um tipo de gráfico de barras que mostra como a frequência de uma variável contínua está distribuída.  As barras agrupam valores em "bins" (intervalos).  É utilizado para analisar a forma da distribuição de uma única variável numérica, como simetria, assimetria, moda e dispersão. O BOXPLOT é um tipo de gráfico que apresenta um resumo de cinco valores: o mínimo, o primeiro quartil (Q1), a mediana (Q2), o terceiro quartil (Q3) e o máximo. É útil tanto para comparar a distribuição de uma variável numérica entre grupos distintos (categorias) quanto para identificar outliers. Um GRÁFICO DE DISPERSÃO é um tipo de diagrama que mostra a relação (correlação) entre duas variáveis numéricas.  Cada ponto simboliza um par de valores (x,y). É útil para observar a conexão (tendência, intensidade e direção) entre duas variáveis, além de identificar agrupamentos ou padrões.

b) A Análise Exploratória de Dados (EDA) é a etapa inicial de qualquer projeto de ciência de dados, consistindo na prática de investigar, resumir e visualizar os principais atributos dos dados, empregando tanto métodos estatísticos (como média, mediana e correlação) quanto representações gráficas (como histogramas, boxplots e gráficos de dispersão).  Seu objetivo principal é analisar a estrutura e os padrões do conjunto de dados, identificar problemas de qualidade (como outliers ou valores ausentes) e orientar a preparação dos dados e a seleção de recursos. Além disso, fornece insights valiosos que ajudam na escolha do algoritmo de machine learning mais adequado, garantindo assim a criação de modelos mais robustos e decisões de negócios mais informadas.

QUESTÃO 2.1:

a) O modelo SUPERVISIONADO é treinado com dados rotulados, formado por pares de entrada (X) e saída desejada (Y).  A meta é compreender um método que relacione a entrada à saída.  É dividido em dois tipos: Regressão (para prever valores contínuos) e Classificação (para prever categorias discretas). Exemplos incluem K-means e DBSCAN. O modelo NÃO SUPERVISIONADO é instruído com dados não rotulados e sem uma variável de saída estabelecida.  O objetivo é identificar a estrutura, os padrões ou os agrupamentos que estão presentes nos dados. Exemplos: K-Means, PCA e Apriori. O modelo de aprendizado POR REFORÇO (agente) aprende a fazer escolhas em um ambiente com o objetivo de aprimorar uma recompensa acumulada.  O aprendizado acontece por meio de tentativas e erros, sem a necessidade de um conjunto de dados estático. Exemplos incluem Q-Learning, SARSA e Deep Q-Networks (DQN).  Aplicado em robótica, desenvolvimento de jogos e sistemas de controle.

b) O Overfitting (ou sobreajuste) ocorre quando o modelo se ajusta muito ao ruído e às características específicas dos dados de treinamento, o que resulta em excelente desempenho no treinamento, mas em baixa capacidade de generalização (alto erro de teste). Técnicas como regularização e validação cruzada ajudam a combater esse problema. Em contraste, o Underfitting (ou subajuste) ocorre quando o modelo é excessivamente simples para captar os padrões fundamentais, o que leva a um desempenho insatisfatório tanto no treinamento quanto no teste (alto erro em ambos). Esse problema é geralmente resolvido aumentando a complexidade do modelo ou adicionando recursos mais relevantes.

c) A Validação Cruzada (Cross-Validation) é uma técnica de reamostragem importante, normalmente realizada no formato K-Fold. Essa técnica analisa a performance e a habilidade de generalização de um modelo de Machine Learning ao dividir o conjunto de dados de treinamento em vários subconjuntos (folds), treinando o modelo várias vezes e utilizando um fold distinto para validação em cada iteração.  Essa metodologia é crucial, pois oferece uma avaliação mais confiável do desempenho do modelo em dados não vistos, contribui para reduzir o overfitting, assegura a utilização eficaz de todos os dados disponíveis e é essencial para a otimização dos hiperparâmetros.

a) 1. Entendimento do Negócio (Business Understanding)

Nessa primeira etapa, busca-se alinhar o objetivo do projeto com as metas do negócio. No caso, a meta é aumentar o valor médio dos pedidos AOV e a taxa de conversão por meio de recomendações personalizadas.
Também é necessário definir onde essas recomendações irão aparecer (por exemplo, na página inicial, na página do produto ou no carrinho de compras) e quais serão os critérios de sucesso. Por fim, devem ser levantados os recursos disponíveis, como orçamento, equipe e infraestrutura.

2. Entendimento dos Dados (Data Understanding)

Aqui ocorre a coleta inicial dos dados, identificando as fontes já existentes, como histórico de pedidos, registros de navegação e informações do catálogo de produtos.
Depois, é feita uma análise exploratória (EDA), que permite compreender melhor a distribuição de compras, a popularidade de determinados itens e o comportamento dos usuários.
É também fundamental avaliar a qualidade dos dados, verificando se estão completos, consistentes e válidos. Por exemplo, pode ser necessário lidar com perfis de usuários inativos ou produtos sem descrição.

3. Preparação dos Dados (Data Preparation)

Nesta fase, os dados passam por um processo de seleção, limpeza e transformação para que possam ser utilizados pelos modelos.
Isso inclui remover registros irrelevantes (como interações de bots ou usuários com poucas ações), tratar valores ausentes e outliers (como pedidos de valor zero), além de construir variáveis úteis, como a matriz de interações entre usuários e produtos.
Por fim, os dados precisam ser formatados de acordo com os requisitos dos algoritmos a serem utilizados.

4. Modelagem (Modeling)

Com os dados preparados, inicia-se a fase de modelagem. Aqui são testadas diferentes técnicas, como filtragem colaborativa (baseada em usuário ou item), fatoração matricial. Os dados são divididos em conjuntos de treino, validação e teste, e os modelos são treinados com ajuste de hiperparâmetros, sempre buscando o melhor desempenho.

5. Avaliação (Evaluation)

Após o treinamento, é preciso avaliar se os resultados atendem às expectativas do negócio. Além das métricas técnicas (precisão, recall, etc.), deve-se verificar se houve impacto em indicadores como CTR ou AOV.
O modelo também é testado em dados nunca utilizados (conjunto de teste), garantindo uma avaliação mais confiável.
Com base nesses resultados, decide-se se o modelo já pode ir para produção ou se há necessidade de retornar a etapas anteriores para melhorias.

6. Implantação (Deployment)

Na fase final, o modelo é integrado ao ambiente de produção, funcionando como um serviço que gera recomendações em tempo real.
Uma prática comum é realizar testes A/B, comparando o novo sistema com a versão anterior (ou com a ausência de recomendações), para medir o impacto real nas métricas do negócio.
Além disso, é fundamental implementar um monitoramento contínuo para acompanhar o desempenho do modelo e garantir que ele continue atendendo às necessidades do sistema ao longo do tempo.

 d)
 1 - Definir métricas de desempenho
Primeiro, precisamos escolher como vamos medir se o modelo está indo bem.
Métricas técnicas: accuracy, precision, recall, F1-score, AUC ou RMSE, dependendo do tipo de modelo.
Métricas de negócio: também é importante olhar para indicadores práticos, como taxa de conversão ou redução de cancelamentos (churn).

2 - Monitorar mudanças nos dados (drift)
O modelo pode perder qualidade se os dados mudarem com o tempo.
Data drift: quando os dados de entrada mudam (ex.: perfil dos clientes).
Concept drift: quando a relação entre entrada e saída muda (ex.: comportamento dos clientes muda).
Existem ferramentas como Evidently AI, River ou Alibi Detect que ajudam a identificar isso.

3 - Registrar previsões e resultados reais
Para saber se o modelo continua bom, é importante guardar o que ele previu e depois comparar com os valores reais (quando estiverem disponíveis).
Isso pode ser feito em bancos de dados ou data lakes com versionamento.

4 - Automatizar alertas
Se o modelo começar a errar mais do que o esperado ou se houver muito drift, é bom ter um sistema de alerta.
Ferramentas como Prometheus + Grafana ou serviços de nuvem como AWS CloudWatch e Azure Monitor podem ajudar nisso.

5 - Dashboard de visualização
É útil ter um painel para acompanhar o desempenho do modelo em tempo real.
Ferramentas como Grafana, Kibana ou Streamlit permitem criar gráficos e dashboards interativos.

6 - Re-treinamento automatizado (opcional)
Se o modelo perder qualidade, podemos acionar um pipeline de re-treinamento com dados novos.
Isso pode ser feito com ferramentas de MLOps, como MLflow, DVC ou Kubeflow.
